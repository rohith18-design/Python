```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("warehouse.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS products (
    product_id INTEGER PRIMARY KEY AUTOINCREMENT,
    product_name TEXT NOT NULL,
    category TEXT,
    quantity INTEGER,
    price REAL
)
""")

conn.commit()

print("Database and table created successfully!")
```

```python
def add_product(name, category, quantity, price):
    cursor.execute("""
    INSERT INTO products (product_name, category, quantity, price)
    VALUES (?, ?, ?, ?)
    """, (name, category, quantity, price))

    conn.commit()
    print("Product added successfully!")
```

```python
add_product("Laptop", "Electronics", 10, 50000)
add_product("Mouse", "Electronics", 25, 500)
add_product("Keyboard", "Electronics", 15, 1200)
```

```python
def view_inventory():
    df = pd.read_sql_query("SELECT * FROM products", conn)
    return df

view_inventory()
```

```python
def search_product(product_name):
    query = """
    SELECT * FROM products
    WHERE product_name LIKE ?
    """
    
    df = pd.read_sql_query(query, conn,
                           params=(f"%{product_name}%",))
    return df
```

```python
search_product("Laptop")
```

```python
def update_quantity(product_id, new_quantity):
    cursor.execute("""
    UPDATE products
    SET quantity = ?
    WHERE product_id = ?
    """, (new_quantity, product_id))

    conn.commit()
    print("Quantity updated successfully!")
```

```python
update_quantity(1, 20)
```

```python
def delete_product(product_id):
    cursor.execute("""
    DELETE FROM products
    WHERE product_id = ?
    """, (product_id,))

    conn.commit()
    print("Product deleted successfully!")
```

```python
delete_product(2)
```

```python
def low_stock(threshold=5):
    query = """
    SELECT * FROM products
    WHERE quantity <= ?
    """

    df = pd.read_sql_query(query, conn,
                           params=(threshold,))
    return df
```

```python
low_stock()
```

```python
def inventory_report():
    total_products = cursor.execute("""
    SELECT COUNT(*) FROM products
    """).fetchone()[0]

    total_quantity = cursor.execute("""
    SELECT SUM(quantity) FROM products
    """).fetchone()[0]

    total_value = cursor.execute("""
    SELECT SUM(quantity * price) FROM products
    """).fetchone()[0]

    print("===== INVENTORY REPORT =====")
    print("Total Products:", total_products)
    print("Total Quantity:", total_quantity)
    print("Total Inventory Value: ₹", total_value)
```

```python
inventory_report()
```

```python
view_inventory()
```

```python
while True:
    print("\n===== WAREHOUSE MANAGEMENT SYSTEM =====")
    print("1. Add Product")
    print("2. View Inventory")
    print("3. Search Product")
    print("4. Update Quantity")
    print("5. Delete Product")
    print("6. Low Stock Report")
    print("7. Inventory Report")
    print("8. Exit")

    choice = input("Enter your choice (1-8): ")

    if choice == "1":
        name = input("Product Name: ")
        category = input("Category: ")
        quantity = int(input("Quantity: "))
        price = float(input("Price: "))
        
        add_product(name, category, quantity, price)

    elif choice == "2":
        print(view_inventory())

    elif choice == "3":
        name = input("Enter product name to search: ")
        print(search_product(name))

    elif choice == "4":
        product_id = int(input("Enter Product ID: "))
        quantity = int(input("Enter New Quantity: "))
        
        update_quantity(product_id, quantity)

    elif choice == "5":
        product_id = int(input("Enter Product ID to delete: "))
        
        delete_product(product_id)

    elif choice == "6":
        threshold = int(input("Enter low stock threshold: "))
        print(low_stock(threshold))

    elif choice == "7":
        inventory_report()

    elif choice == "8":
        print("Exiting Warehouse Management System...")
        break

    else:
        print("Invalid choice! Please try again.")
```
